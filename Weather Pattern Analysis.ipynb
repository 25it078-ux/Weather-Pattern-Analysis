```python
# Weather Pattern Analysis

import pandas as pd
import matplotlib.pyplot as plt

# Sample Weather Dataset
data = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'Temperature (°C)': [30, 32, 31, 33, 35, 34, 32],
    'Humidity (%)': [65, 60, 62, 58, 55, 57, 61],
    'Rainfall (mm)': [5, 0, 2, 0, 0, 1, 4]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display Dataset
print("Weather Dataset:\n")
print(df)

# Basic Statistics
print("\nWeather Statistics:\n")
print(df.describe())

# Average Temperature
avg_temp = df['Temperature (°C)'].mean()
print(f"\nAverage Temperature: {avg_temp:.2f} °C")

# Highest Temperature
max_temp = df['Temperature (°C)'].max()
print(f"Highest Temperature: {max_temp} °C")

# Line Chart - Temperature Trend
plt.figure(figsize=(8,5))
plt.plot(df['Day'], df['Temperature (°C)'], marker='o')
plt.title('Temperature Trend')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.show()

# Line Chart - Humidity Trend
plt.figure(figsize=(8,5))
plt.plot(df['Day'], df['Humidity (%)'], marker='o')
plt.title('Humidity Trend')
plt.xlabel('Day')
plt.ylabel('Humidity (%)')
plt.grid(True)
plt.show()

# Bar Chart - Rainfall
plt.figure(figsize=(8,5))
plt.bar(df['Day'], df['Rainfall (mm)'])
plt.title('Rainfall Analysis')
plt.xlabel('Day')
plt.ylabel('Rainfall (mm)')
plt.show()

# Combined Weather Parameters
plt.figure(figsize=(10,5))
plt.plot(df['Day'], df['Temperature (°C)'], marker='o', label='Temperature')
plt.plot(df['Day'], df['Humidity (%)'], marker='s', label='Humidity')
plt.title('Weather Pattern Analysis')
plt.xlabel('Day')
plt.ylabel('Values')
plt.legend()
plt.grid(True)
plt.show()

# Insights
print("\nWeather Insights:")
print(f"- Average Temperature: {avg_temp:.2f} °C")
print(f"- Highest Temperature Recorded: {max_temp} °C")
print("- Weather trends have been visualized using charts.")
```
